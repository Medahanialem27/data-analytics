## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_3 = np.random.random(size=(2, 1))
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random chices from an Array

```python
students = np.array([
    "Tim1", "Abdul2", "Israel3", "Arash4",
    "Faiz5", "Cesar6", "Earl7", "Meric8",
    "Maream9", "Matthew10", "Stephanie11", "Alizon12"
])
```

```python
ca11_on = np.random.choice(students)
print(ca11_on)
```

```python
breakout_groups = np.random.choice(students,size=(3, 4))
print(breakout_groups)
```

```python
# Not ideal for groups:choice repeats by defaul, so some students may be selected than once.
```

```python
ca11_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0, 0.30, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10,  0.05, 0.05, 0, 0]
)
print(ca11_on_weighted)
```

```python
# probabilities follow the students order:Tim1 0%, Abdul12 30%, Israel3 through Meric8 10% each, Maream9and Metthew10 5% each, Stephanie11 and Alizon12 0%.
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
