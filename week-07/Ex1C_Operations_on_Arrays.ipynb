# Sample Arrays

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The rows were shuffled into a different order, but the values within each row stayed together.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

Use shuffle when we want to rearrange the original array. Use permutation when we want a shuffled copy and want to keep the orginal array unchanged.

## Mathematical Operations

```python
print(r_1)
print(r_1.shape)
print(r_2)
print(r_2.shape)

result_1 = np.add(r_1, r_2)
print(result_1)
print(result_1.shape)
```

```python
print(r_1)
print(r_1.shape)
print(r_3)
print(r_3.shape)

result_2 = np.add(r_1, r_3)
print(result_2)
print(result_2.shape)
```

```python
print(r_1)
print(r_1.shape)
print(r_4)
print(r_4.shape)

# Numpy add 1 to every value in r_4
print(np.add(r_4, np.array([1])))
```

```python
print(r_2.shape)
print(r_4.shape)

# These shapes cannot be broadcast because 2 and 5 columns differ.
np.add(r_2, r_4)
```

```python
print(np.add(r_4, np.array([1])))
```

```python
print(np.add(r_4, np.array([[1]])))
```

```python
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

```python
# This should raise an error because 2 rows cannot match 4 rows.
print(np.add(r_4, np.array([[1], [1]])))
```

Broadcasting lets Numpy combine array of different shapes when their dimensions are compatible. with np.add, matching values are added.

## Grocery problem

```python
bag1 = np.array([2 ,3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
print(f"The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.")
```

```python
totals = np.add(bag1, bag2)
print(f"There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
ben = np.array([0, 5, 5, 3, 4])
carla = np.array([3, 6, 0, 3, 4])

all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
print(np.sum(all_knocks, axis=1))
print(np.sum(all_knocks))
```

```python
daily_knocks = np.sum(all_knocks, axis=0)
rolling_votes = np.cumsum(daily_knocks)* 0.5
print(rolling_votes)
```
