## Looping Through Arrays

```python
arrA = np.array([[11, 12,  13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

# The first loop goes through the array one row at a time,so print two rows. The second loop goes through each number inside each row, so it prints six individual numbers.

```python
for x in arrA:
    for y in x:
        loop_result = y * 2

print(loop_result)
        # Loop_result is overwritten each time so it keeps only the last result: 32.
```
