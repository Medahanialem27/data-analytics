# Exercise 2.A. Using built-in functions
# Lab 2. Parsing strings - part code

```python
def get_supplier_code(part_code):
    colon_position = part_code.index(':')
    return part_code[:colon_position]
```

```python
def get_product_num(part_code):
    colon_position = part_code.index(':')
    hyphen_position = part_code.index('-')
    return part_code[colon_position + 1:hyphen_position]
    
```

```python
def get_size(part_code):
    hyphen_position = part_code.index('-')
    return part_code[hyphen_position + 1:]
```

```python
part_code1 = 'ACME:123-L'
part_code2 = 'DI:12345-M'
part_code3 = 'ACE:1-12'
```

```python
# Test the functions with the first part code
print(f"For {part_code1}, the supplier is {get_supplier_code(part_code1)}. part number {get_product_num(part_code1)} is available in size {get_size(part_code1)}.")
```

```python
# Test the functions with the second part code
print(f"For {part_code2}, the supplier is {get_supplier_code(part_code2)}. part number {get_product_num(part_code2)} is available in size {get_size(part_code2)}.")
```
