# Exercise 2.B, Lab 2: Times and dates
This lab practices working with dates and times using Python's datetime function

```python
import datetime
# Today
```

```python
today = datetime.datetime.now()
```

```python
today_format = today.strftime("%A, %B %d, %Y")
print(f"Today is {today_format}")
```

# My birthday

```python
birthday = datetime.date(2000, 1, 15)
```

```python
birthday_format = birthday.strftime("%m/%d/%Y")
print(f"My birthday is {birthday_format}")
```

# 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
ninety_d_format = ninety_d.strftime("%B %d, %Y")
print(f"90 days from today is {ninety_d_format}")
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
```

```python
dinner_time_format = dinner_time.strftime("%I:%M %P")
print(f"let's meet for dinner at {dinner_time_format}")
```
