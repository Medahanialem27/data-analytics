# Exercise 2.C Creating and Calling Functions
## Lab 1
This lab practices defining and calling custom python functions.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nmy favorite movie is {movie}.\nmy favorite musician or band is {music}"

print(favorite_things("Netsanet", "Forever my Girl", "Rihanna"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Year Up United Data Analyst program to build my skills and prepare for a data analytics career."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Oakland", "it feels like home"))
```

```python
print(favorite_place("Hawaii", "I enjoy the beaches"))
```
